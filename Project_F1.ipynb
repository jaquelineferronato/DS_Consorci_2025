```python


import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression


file_path = 'Tabela_nova_podium_paises.csv'
data = pd.read_csv("C:/Users/ferronat/Desktop/JF/F1/Tabela_nova_podium_paises.csv")


dados = data[['Year', 'Constructor', 'Position']]
dados = dados[dados['Position'] <= 3]


podios_ano_construtor = dados.groupby(['Year', 'Constructor']).size().reset_index(name='Podiums')


construtores = podios_ano_construtor['Constructor'].unique()


previsoes = pd.DataFrame()


ultimo_ano = data['Year'].max()
anos_previsao = np.arange(ultimo_ano + 1, ultimo_ano + 6)


for construtor in construtores:
   
    df_construtor = podios_ano_construtor[podios_ano_construtor['Constructor'] == construtor]
    X = df_construtor['Year'].values.reshape(-1, 1)
    y = df_construtor['Podiums'].values

    
    if len(X) >= 2:
        model = LinearRegression()
        model.fit(X, y)
        previsoes_construtor = model.predict(anos_previsao.reshape(-1, 1))
        df_previsao = pd.DataFrame({'Year': anos_previsao, 'Constructor': construtor, 'Predicted_Podiums': previsoes_construtor})
        previsoes = pd.concat([previsoes, df_previsao], ignore_index=True)

 
previsoes.to_csv('Previsao_Podiums_Construtores.csv', index=False)

print('Previsão concluída e salva em Previsao_Podiums_Construtores.csv')
```

```python


import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression


file_path = 'Tabela_nova_podium_paises.csv'
data = pd.read_csv("C:/Users/ferronat/Desktop/JF/F1/Tabela_nova_podium_paises.csv")


dados = data[['Year', 'Constructor', 'Position']]
dados = dados[dados['Position'] <= 3]


podios_ano_construtor = dados.groupby(['Year', 'Constructor']).size().reset_index(name='Podiums')


ultimo_ano = data['Year'].max()
anos_ultimos_5 = list(range(ultimo_ano - 4, ultimo_ano + 1))
construtores_ativos = podios_ano_construtor[podios_ano_construtor['Year'].isin(anos_ultimos_5)]['Constructor'].unique()


previsoes = pd.DataFrame()


anos_previsao = np.arange(ultimo_ano + 1, ultimo_ano + 6)


for construtor in construtores_ativos:
    
    df_construtor = podios_ano_construtor[podios_ano_construtor['Constructor'] == construtor]
    X = df_construtor['Year'].values.reshape(-1, 1)
    y = df_construtor['Podiums'].values

  
    if len(X) >= 2:
        model = LinearRegression()
        model.fit(X, y)
        previsoes_construtor = model.predict(anos_previsao.reshape(-1, 1))
        
        previsoes_construtor = np.maximum(previsoes_construtor, 0)
        df_previsao = pd.DataFrame({'Year': anos_previsao, 'Constructor': construtor, 'Predicted_Podiums': previsoes_construtor})
        previsoes = pd.concat([previsoes, df_previsao], ignore_index=True)


previsoes.to_csv('Previsao_Podiums_Construtores.csv', index=False)

print('Previsão concluída e salva em Previsao_Podiums_Construtores.csv')
```
